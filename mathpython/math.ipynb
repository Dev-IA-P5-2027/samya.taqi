# dérivation

```python
from sympy import symbols, Derivative

x = symbols('x')
expr = x**2 + 3*x + 2
 
# Calcul de la dérivée de l'expression par rapport à x
derivative_expr = Derivative(expr, x)
 
# Affichage de la dérivée
print(derivative_expr)
```

```python
#la loi normale 
import scipy.stats as  stats 
from scipy.stats import norm

moyenne=178
sigma=10
valeur= 3
X=stats.norm(moyenne, sigma)
proba= X.cdf(moyenne+valeur)-X.cdf(moyenne-valeur)
print(proba)
```

# Dérivé

```python
import sympy as sp

#Définir la variable
x=sp.symbols('x')

#Définir la fonction
f=x*sp.log(x)

#calcule de dérivé
f_prime=sp.diff(f,x)
print(f_prime)

#calculer f'(3)
valeur=f_prime.subs(x,3)
print(valeur)
print(valeur.evalf())
```

# loi binomiale

```python
from math import *
import numpy as np
from scipy.stats import binom
n=80
p=0.02192
k=5

a=binom.pmf(k, n, p)   # P(X = k)
b=binom.cdf(k, n, p)   # P(X ≤ k)
c=binom.sf(k, n, p)    # P(X > k)
d=binom.mean(n, p)     # Espérance = n*p
e=binom.var(n, p)      # Variance = n*p*(1-p)
 

print(a ,b,c,d,e)
```
