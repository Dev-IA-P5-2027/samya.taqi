**en français**

```python
import re

contractions_fr = {
    "j'": "je ",
    "t'": "te ",
    "c'": "ce ",
    "d'": "de ",
    "l'": "le ",
    "qu'": "que ",
    "n'": "ne ",
    "s'": "se "
}

def fix_contractions_fr(text):
    for contr, full in contractions_fr.items():
        text = re.sub(rf"\b{contr}", full, text)
    return text

# Exemple
text = "J'aime le chocolat et c'est délicieux."
print(fix_contractions_fr(text))
```

# Standardising accent characters

```python
import unicodedata

def standardize_acceted_chars(text):
    return unicodedata.normalize('NFKD',text).encode('ascii','ignore').decode('utf-8', 'ignore')
```

```python
input_text= 'Sómě words as résumé, café ,prótest, divorcé , cordinate, exposé , latté .'
print(standardize_acceted_chars(input_text))
```

```python
def remove_html_tags_regex(text):
    """Remove HTML tags using regular expression."""
    clean= re.compile('<.*?>')
    return re.sub(clean, '',text)
```

```python
html_text="<p>Hello , <b>world</b>!</p> <a href='#'>Click here</a>"
clean_text= remove_html_tags_regex(html_text)
print(clean_text)
```

```python
#removing numbers/digits

def remove_numbers(text):
    result=re.sub(r'\d{7,}', '<phone number>', text)
    return result
```

**convert the numbers into words**

```python
import inflect
p=inflect.engine()
#CONVERTIE LE NUMERO EN MOTS
def convert_numbre(text):
    #SPLIT STRING INTO LIST OF WORDS 
    temp_str=text.split()
    # initialiser une liste vide 
    new_string=[]
    for word in temp_str:
        # if le mot is a digit, convertie the digit
        if word.isdigit():
            temp=p.number_to_words(word)
            new_string.append(temp)
            #appeler le mot as it is
        else:
            new_string.append(word)
            #join the words of new_string to form a string
    temp_str= ' '.join(new_string)
    return temp_str
input_text= 'There are 3 balls in this bag, and 12 in the other one.'
convert_numbre(input_text)
```

```python
def remove_punctuation(text, punct_text):
    for punc in punct_text:
        if punc in text:
            text=text.replace(punc, ' ')
    return text.strip()

remove_punctuation("Hey @Lisa, did you know thet the summer break is coming? Amazing right !! It's only 5 more days !!", regular_punct)
```

```python
# gérer emoji/emoticon pour analyser sentiment 
def remove_emojis_regex(text):
    """Remove emojis using Unicode emoji ranges."""
    #unicode ranges for emojis
    emoji_pattern= re.compile("["
                              u"\U0001F600-\U0001F64F"
                              u"\U0001F300-\U0001F5FF"
                              u"\U0001F680-\U0001F6FF"
                              u"\U0001F1E0-\U0001F1FF"
                              u"\U00002702-\U000027B0"
                              u"\U000024C2-\U0001F251"
                              u"\U0001f926-\U0001f937"
                              u"\U00010000-\U0010ffff"
                                "]+", flags=re.UNICODE)
    
    return emoji_pattern.sub(r'', text)
```

```python
text="Hello! How are yo?🙂​ I'm in NYC"
clean_text=remove_emojis_regex(text)
print(clean_text)
```

```python
import emoji

def remove_emojis_emoji_library(text):
    """Remove emojis using the emoji library."""
    return emoji.replace_emoji(text, replace='')

def replace_emojis_with_text(text):
    """Replace emojis with their text descriptions using the emoji library"""
    return emoji.demojize(text)
 
text="Hello! 😊👋 how are you? i'm in Morrocco ❤️ "
print(remove_emojis_emoji_library(text))
print(replace_emojis_with_text(text))
    
```

```python
url_pattern=re.compile(r'https://\S+')
def remove_url(text):
    return url_pattern.sub('', text)
text="Visit https://www.google.com for more information."
remove_url(text)
```

```python
url_pattern=[]
url_pattern.append(re.compile(r'https:?//\S+'))
url_pattern.append(re.compile(r'http:?//\S+'))
url_pattern.append(re.compile(r'www.\S+'))

def remove_urls2(text):
    for pattern in url_pattern:
        text=pattern.sub('<this link>', text)
    return text

input_text="Visite the folowing link: www.google.com"
print(remove_urls2(input_text))
```

```python
# the required libraries
from nltk.tokenize import word_tokenize
from nltk.tokenize import word_tokenize

def remove_stopwords(text):
    stop_words=set(stopwords.words("english"))
    word_tokens= word_tokenize(text)

    filtred_text=[word for word in word_tokens if word.lower() not in stop_words]
    return filtred_text 


exemple_text="This is a sample sentence and we are going to remove the stopwords from this."

print("Before removing stopwords:", exemple_text)
print("After removieng stopwords", remove_stopwords(exemple_text))
```

**porte stemmer algorithm**

```python
from nltk.stem.porter import PorterStemmer
from nltk.tokenize import word_tokenize

stemmer= PorterStemmer()
def stem_words(text):
    word_tokens= word_tokenize(text)
    stems= [stemmer.stem(word) for word in word_tokens]
    return stems

text="data science uses scientific methods algorithms and many types of processes"
stem_words(text)
```

```python
#lemmatization using nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

lemmatizer= WordNetLemmatizer()

def lemma_words(text):
    words_tokens=word_tokenize(text)
    lemmas=[lemmatizer.lemmatize(word) for word in words_tokens]
    return lemmas

text='The quick brown foxes are jumping over the lazy dogs.'
print(lemma_words(text))
```
