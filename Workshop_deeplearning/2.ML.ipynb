### **Exercice 3 :** Classification avec SVM
Transformer la qualité du vin en une variable binaire et appliquer un modèle de classification.

### Instructions
1. Convertissez `quality` en une variable binaire : `1` (qualité >= 6) et `0` (qualité < 6).
2. Entraînez un modèle SVM sur l’ensemble train/test.
3. Évaluez la performance du modèle avec `accuracy_score`.

```python
df["quality"] = df["quality"].apply(lambda x: 1 if x >= 6 else 0)
df
```

```python
# Réponse 
if "quality" in df.columns and df["quality"] >= 6:
    df["quality"] = 1
else:
    df["quality"] = 0
df
```

```python
df_clss=df
for i in range(len(df)):

    if df_clss["quality"][i]>= 6:
        df_clss.loc[i,"quality"]= 1
    else:
        df_clss.loc[i,"quality"] = 0
df_clss
```

```python
df_clss["quality"].value_counts()
```

```python
x= df_clss.drop("quality", axis=1)
y=df_clss["quality"]
```

```python
x_train, x_test, y_train, y_test= train_test_split(x, y, test_size=0.2 , random_state=42)
```

```python
model_svm= svm.SVC(kernel="linear", C=1.0, random_state=42)
model_svm.fit(x_train, y_train)
```

```python
y_predict_svm= model_svm.predict(x_test)
accuracy_svm= accuracy_score(y_test, y_predict_svm)
print("__SVM__")
print("Accuracy =", accuracy_svm)
matrice= confusion_matrix(y_test, y_predict_svm)

print(matrice)
```
