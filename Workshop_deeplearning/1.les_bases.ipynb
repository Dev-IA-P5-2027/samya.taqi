### **Exercice 1:** Implémentez une fonction qui calcule la somme des nombres d'une liste.

```python
# Réponse :
def somme_nombre_liste(liste):
    somme = 0
    for nombre in liste:
        somme += nombre
    return somme

resultat = somme_nombre_liste([1,2,3,4])
print(resultat)
```

### **Exercice 2:** Écrivez une fonction qui retourne la moyenne des valeurs d'une liste.

```python
# Réponse :
def moyenne_liste(liste):
    if len(liste) == 0:
        return 0
    else:
        return sum(liste) / len(liste)
print(moyenne_liste([1,2]))
```

```python
def somme_liste(liste):
    return sum(liste)

print(somme_liste([1,2,3,5]))
```

### **Exercice 3:** Implémentez une fonction qui trouve le maximum d'une liste sans utiliser `max()`.

```python
# Réponse :
def maximum_liste(liste):
    if len(liste) == 0:
        return None
    else:
        return max(liste)
print(maximum_liste([1,2,3,4,5]))
```

```python
from numpy import sort
def found_max(nb_list):
    if(len(nb_list) < 1):
        return 0
    return sort(nb_list)[-1]
 
def found_max_2(nb_list):
    if(len(nb_list) < 1):
        return 0
    current_max = None
    for value in nb_list:
        if(current_max is None
           or current_max < value):
            current_max = value
    return current_max
found_max([1,2,3])
```

### **Exercice 4:** Implémentez une fonction qui regroupe les éléments d'une liste en fonction de leur parité (pairs et impairs).

```python
# Réponse :
def regroupe_liste_parité(liste):
    pairs = []
    impairs = []
    for nombre in liste:
        if nombre % 2 == 0:
            pairs.append(nombre)
        else:
            impairs.append(nombre)
    return pairs, impairs
pairs, impairs = regroupe_liste_parité([1,2,3,4,5,6,12,15,20,25])
print("Pairs:", pairs)
print("les nombrse impairs:", impairs)
```
